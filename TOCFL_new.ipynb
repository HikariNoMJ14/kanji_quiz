```python
import pandas as pd

mapping = {
    'Novice 1': 1, 'Novice 2': 1,
    'A Level 1': 2, 'A Level 2': 3,
    'B Level 1': 4, 'B Level 2': 5, 'C Level 1': 6
}


def replace(x):
    return mapping[x]
```

```python
tocfl = pd.read_csv('./data/TOCFL_en.csv', index_col=0)
tocfl['level'] = tocfl['level'].apply(replace)
tocfl = tocfl.drop(['level 2'], axis=1)
tocfl = tocfl.reset_index()
tocfl['index'] = tocfl['index'] + 1
tocfl['number'] = tocfl['number'].astype(int)
tocfl['char 2'] = (tocfl['character']
                   .str.replace(r'\(.*\)', '', regex=True)
                   .str.replace('（.*）', '', regex=True)
                   .str.strip())
tocfl['char 3'] = (tocfl['char 2']
                   .str.replace(r'\/.*', '', regex=True)
                   .str.strip())
tocfl['zhuyin'] = tocfl['zhuyin'].str.replace('', '')
```
